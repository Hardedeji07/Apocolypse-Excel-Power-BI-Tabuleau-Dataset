# Iris Dataset Analysis

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target

# Map target integers to species names for better readability
df['species'] = df['species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

print("First 5 rows of the dataset:")
display(df.head())
```

## Data Cleaning and Basic Statistical Analysis

```python
# Check for missing values
print("\nMissing values per column:")
display(df.isnull().sum())

# Display basic statistical summary
print("\nDescriptive statistics of the dataset:")
display(df.describe(include='all'))

# Display information about the DataFrame (data types, non-null counts)
print("\nDataFrame Info:")
df.info()
```

## Data Visualization

```python
# Pair plot to visualize relationships between all features, colored by species
sns.pairplot(df, hue='species', palette='viridis')
plt.suptitle('Pair Plot of Iris Features by Species', y=1.02)
plt.show()
```

```python
# Distribution of each feature using histograms and KDE plots
plt.figure(figsize=(15, 10))
for i, feature in enumerate(iris.feature_names):
    plt.subplot(2, 2, i + 1)
    sns.histplot(df[feature], kde=True)
    plt.title(f'Distribution of {feature}')
plt.tight_layout()
plt.show()
```

```python
# Box plots for each feature by species to see variations
plt.figure(figsize=(15, 10))
for i, feature in enumerate(iris.feature_names):
    plt.subplot(2, 2, i + 1)
    sns.boxplot(x='species', y=feature, data=df, hue='species', palette='muted', legend=False)
    plt.title(f'{feature} by Species')
plt.tight_layout()
plt.show()
```

```python
# Correlation matrix heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(df[iris.feature_names].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Iris Features')
plt.show()
```
