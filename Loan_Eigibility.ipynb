```python
import pandas as pd
import os

# Using the provided file path
file_path = '/Loan_Eligibility_Data.csv'

if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    print("Dataset loaded successfully!")
    display(df.head())
else:
    print(f"Error: '{file_path}' not found. Please ensure the file is uploaded to the root directory.")
    df = None
```

### Dataset Structure and Features

```python
if df is not None:
    # Display the first 5 rows of the dataset
    print("First 5 rows of the dataset:")
    display(df.head())

    # Get a concise summary of the DataFrame
    print("\nDataFrame Info:")
    df.info()

    # Generate descriptive statistics
    print("\nDescriptive Statistics:")
    display(df.describe(include='all'))
```

### Target Variable Distribution (Loan Status)

```python
if df is not None:
    # Check the distribution of the target variable 'Loan_Status'
    if 'Loan_Status' in df.columns:
        print("\nDistribution of 'Loan_Status':")
        display(df['Loan_Status'].value_counts())
        display(df['Loan_Status'].value_counts(normalize=True) * 100)
    else:
        print("Warning: 'Loan_Status' column not found in the dataset.")
```

### Correlation Analysis of Numerical Features

```python
import matplotlib.pyplot as plt
import seaborn as sns

if df is not None:
    # Selecting only numerical columns for correlation analysis
    numerical_cols = df.select_dtypes(include=['float64', 'int64']).columns
    corr_matrix = df[numerical_cols].corr()

    print("Correlation Matrix:")
    display(corr_matrix)

    # Plotting the heatmap
    plt.figure(figsize=(10, 6))
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
    plt.title('Correlation Heatmap of Numerical Features')
    plt.show()
```
